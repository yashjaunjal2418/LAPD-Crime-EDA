1. VICTIM AGE

```python
bins = [0, 20, 40, 60, 80, 100]
labels = ["0-20", "21-40", "41-60", "61-80", "81-100"]

df["Age Group"] = pd.cut(df["Vict Age"], bins=bins, labels=labels)

counts = df["Age Group"].value_counts().sort_index()

plt.figure(figsize=(9, 5))
plt.bar(counts.index, counts.values)

for i, value in enumerate(counts.values):
    plt.text(i, value, str(value), ha="center")

plt.title("Victim Age Group Distribution")
plt.xlabel("Age Groups")
plt.ylabel("Number of Victims")

plt.show()
```

#In this 21 to 40 age group has higher crime rate.
#reason behind it is making connections with local gangs, assault with partners, urge to make money faster gets involves in car theft , business frauds , property crimes.

3. CRIME BY HOURS

```python
df["Hour"] = df["TIME OCC"] // 100
counts = df.groupby("Hour")["DR_NO"].count()
plt.figure(figsize=(10, 5))
plt.bar(counts.index, counts.values, width=0.5)
for i, value in enumerate(counts.values):
    plt.text(i, value, str(value), ha="center",fontsize=7)

plt.title("Crime by Hour")
plt.xlabel("Hour (0-23)")
plt.ylabel("Number of Crimes")

plt.xticks(range(24))
plt.show()
```

#12pm has the highest crime rate because 
#Crowded public places may create opportunities for theft and other crimes.
#Shops, banks and commercial areas are active during the afternoon.
#Many theft occur during this period while People are travelling, shopping, working and doing daily activities.

4. TOP 20 CRIMES BY POLICE AREA

```python
counts = df.groupby("AREA NAME")["DR_NO"].count()
counts = counts.sort_values(ascending=False).head(20)
plt.figure(figsize=(12, 7))
plt.barh(counts.index, counts.values)
for i, value in enumerate(counts.values):
    plt.text(value, i, str(value), va="center")

plt.title("Top 20 Crime by Police Area")
plt.xlabel("Number of Crimes")
plt.ylabel("Police Area")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

#Here 77th street has the highest crime rate.
#It's because it is a populated area.

3. CRIME BY HOUR IN WEEKDAYS

```python

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

data = df.pivot_table(
    index="Day", columns="Hour",
    values="DR_NO", aggfunc="count"
).fillna(0)

data = data.reindex(days)

plt.figure(figsize=(12, 5))
plt.imshow(data, cmap="Reds", aspect="auto")

plt.colorbar(label="Crime Count")
plt.xlabel("Hour")
plt.ylabel("Day")
plt.yticks(range(7),days)
plt.xticks(range(0,24))
plt.title("Crime by Hour and Day")
plt.show()
```

#This heatmap analyzes crime distribution by day and hour. It helps us identify peak crime periods, 
understand weekly patterns, and find the days and times when more crimes are reported as occurring.

#Uses of This Heatmap:
Helps identify peak crime hours and days.
Helps understand weekly crime patterns.
Helps authorities study when additional resources may be needed.
Helps researchers investigate possible time-related crime patterns.

1. Victim patterns across age, sex and crime

```python
import pandas as pd
import matplotlib.pyplot as plt

df['DATE OCC'] = pd.to_datetime(df['DATE OCC'])

df['Year'] = df['DATE OCC'].dt.year
df['Month'] = df['DATE OCC'].dt.month

data = df.groupby(['Year', 'Month']).size().unstack(fill_value=0)

plt.imshow(data, cmap='YlOrRd', aspect='auto')

plt.xticks(range(12), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
                       'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])

plt.yticks(range(len(data)), data.index)

plt.colorbar(label='Crime Count')
plt.title('Year vs Month Crime Count')
plt.xlabel('Month')
plt.ylabel('Year')

plt.show()
```

#This heatmap represents crime distribution across different months and years. 
#It helps us identify high-crime and low-crime months, compare yearly patterns, and understand seasonal variations in recorded crime.

#In year 2016 has more crime count compare to other years.

#Use of this HEATMAP:
To identify monthly and yearly crime patterns.
To understand seasonal crime trends.
To identify months that may need further investigation.
To help authorities plan resource allocation based on historical patterns.

2. Male and female victim trends over years

```python
df['Year'] = pd.to_datetime(df['DATE OCC']).dt.year

data = df.groupby(['Year', 'Vict Sex']).size().unstack(fill_value=0)

data[['Male', 'Female']].plot(marker='o')

plt.title('Victim Count by Year and Sex')
plt.xlabel('Year')
plt.ylabel('Victim Count')
plt.xticks(range(2010,2020))
plt.legend(['Male', 'Female'])
plt.grid()
plt.show()
```

#This line chart compares male and female victim counts from 2010 to 2019. It helps us understand gender-wise victim patterns, compare yearly victim counts, and identify changes in victim trends over time.

3. CRIME TRENDS DIFFER BY AREA THROUGH YEARS

```python
df['Year'] = pd.to_datetime(df['DATE OCC']).dt.year

data = df.groupby(['Year', 'AREA NAME']).size().unstack()

data.plot(kind='line', marker='o', figsize=(12, 6))

plt.title('Crime Trends by Area Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Crimes')
plt.xticks(range(2010,2020))
plt.legend(title='Area',loc='upper left', bbox_to_anchor=(1, 1))
plt.grid(True)

plt.show()
```

#This graph helps us compare yearly crime trends across different areas of Los Angeles and identify areas where crime counts are increasing or decreasing.
#It helps us identify high-crime areas, observe increasing or decreasing trends, and understand how crime patterns differ from one area to another.
#If the line for 77th Street is higher than the other areas, it means 77th Street has more recorded crimes in those years.
